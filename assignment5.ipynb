
QUES:18 How would you implement a basic logging setup that logs to a file with rotation after 1MB?

```python
import logging
from logging.handlers import RotatingFileHandler

try:
    # Create logger
    logger = logging.getLogger("my_logger")
    logger.setLevel(logging.DEBUG)   # capture all levels
    
    # Create a rotating file handler
    handler = RotatingFileHandler(
        "app.log",        # log file name
        maxBytes=1*1024*1024,  # 1 MB
        backupCount=5         # keep last 5 rotated logs
    )

    # Define log format
    formatter = logging.Formatter("%(asctime)s - %(levelname)s - %(message)s")
    handler.setFormatter(formatter)

    # Add handler to logger
    logger.addHandler(handler)

    # Example usage
    for i in range(10000):
        logger.info(f"Logging entry number {i}")

    print("Logging setup successful. Logs are being written to 'app.log'.")

except PermissionError:
    print("Error: You don’t have permission to write to the log file.")

except FileNotFoundError:
    print("Error: Log file path not found.")

except Exception as e:
    print("Unexpected error occurred while setting up logging:", e)
```

QUES:21 Write a Python program that reads a file and prints the number of occurrences of a specific word ?

```python


def count_word_in_file(filename, word):
    try:
        with open(filename, "r") as file:
            contents = file.read().lower()   # convert to lowercase for case-insensitive match
            word = word.lower()
            count = contents.split().count(word)
            print(f"The word '{word}' occurs {count} times in '{filename}'.")

    except FileNotFoundError:
        print("Error: File not found.")

    except PermissionError:
        print("Error: You don’t have permission to read this file.")

    except Exception as e:
        print("An unexpected error occurred:", e)

filename = "file1.txt"
word_to_search = "python" 
count_word_in_file(filename, word_to_search)
```

QUES:22  How can you check if a file is empty before attempting to read its contents?

```python
import os

filename = "example.txt"

try:
    if os.path.getsize(filename) == 0:
        print("The file is empty.")
    else:
        with open(filename, "r") as file:
            contents = file.read()
            print("File contents:\n", contents)

except FileNotFoundError:
    print("Error: The file does not exist.")
except Exception as e:
    print("Unexpected error:", e)
```
