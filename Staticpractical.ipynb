1. What is hypothesis testing in statistics?

Hypothesis testing is a statistical method used to make decisions or draw conclusions about a population based on sample data. It helps determine whether there is enough evidence to reject a given assumption.

2. What is the null hypothesis, and how does it differ from the alternative hypothesis?

 -Null hypothesis (H₀): Assumes no effect, no difference, or no relationship.
 -Alternative hypothesis (H₁ or Hₐ): Assumes there is an effect, difference, or relationship.

3. What is the significance level in hypothesis testing, and why is it important?

The significance level (α) is the probability of rejecting the null hypothesis when it is actually true (Type 1 error).
Common values: 0.05 or 0.01.

It sets the threshold for decision-making.

4. What does a P-value represent in hypothesis testing?

The P-value is the probability of obtaining results as extreme as the observed results, assuming the null hypothesis is true.

5. How do you interpret the P-value in hypothesis testing?

- If P ≤ α → Reject H₀

- If P > α → Fail to reject H₀

Smaller P-value = stronger evidence against H₀.

6. What are Type 1 and Type 2 errors in hypothesis testing?

- Type 1 error: Rejecting a true null hypothesis

- Type 2 error: Failing to reject a false null hypothesis

7. What is the difference between a one-tailed and a two-tailed test?

- One-tailed test: Tests effect in one direction

- Two-tailed test: Tests effect in both directions

8. What is the Z-test, and when is it used?

A Z-test is used to test population means when:

- Sample size is large (n ≥ 30)

- Population standard deviation is known

- Data is normally distributed

9. How do you calculate the Z-score, and what does it represent?


           Z= X-μ/ σ

It represents how many standard deviations a value is from the mean.  


10. What is the T-distribution, and when should it be used?

The T-distribution is similar to the normal distribution but has heavier tails.

It is used when:

- Sample size is small (n < 30)

- Population standard deviation is unknown

11. What is the difference between a Z-test and a T-test?


Z-test	                                         T-test
Large sample	                                 Small sample
σ known	                                         σ unknown
Normal distribution	                             T-distribution

12. What is the T-test, and how is it used?

A T-test compares means to determine if they are significantly different.

Types:

- One-sample T-test
- Independent T-test
- Paired T-test


13. What is the relationship between Z-test and T-test?

Both are used to compare means.

As sample size increases, T-test approaches Z-test.

14. What is a confidence interval?

A confidence interval is a range of values that likely contains the population parameter with a specified confidence level (e.g., 95%).


15. What is the margin of error?


Margin of error shows how much the estimate may differ from the true value.

Smaller margin = more precise estimate.

16. How is Bayes' Theorem used in statistics?

Bayes’ Theorem updates probability based on new evidence.
Used in:

- Machine learning

- Medical diagnosis

- Risk prediction


17. What is the Chi-square distribution, and when is it used?

Chi-square distribution is used for categorical data to test relationships between variables.

18. What is the Chi-square goodness of fit test?

It checks whether observed data matches an expected distribution.

19. What is the F-distribution, and when is it used?

F-distribution is used to compare variances and in tests like ANOVA.

20. What is an ANOVA test, and its assumptions?

ANOVA (Analysis of Variance) tests differences among means of multiple groups.

Assumptions:

- Normality

- Equal variances

- Independence


21. What are the different types of ANOVA tests?

- One-way ANOVA

- Two-way ANOVA

- Repeated measures ANOVA


22. What is the F-test, and how does it relate to hypothesis testing?

The F-test compares variances and is the basis of ANOVA.

It helps determine whether group means differ significantly.

```python
# QUES:3 Write a Python function to calculate the probability distribution function (PDF) of a Bernoulli distribution
def bernoulli_pdf(x, p):
    if x == 0:
        return 1 - p
    elif x == 1:
        return p
    else:
        return 0

print("P(X=1):", bernoulli_pdf(1, 0.7))
print("P(X=0):", bernoulli_pdf(0, 0.7))
```

```python
# QUES:1 Write a Python program to perform a Z-test for comparing a sample mean to a known population mean and interpret the results
import numpy as np
from math import sqrt

sample = np.array([52, 55, 50, 48, 54, 51])
pop_mean = 50
pop_std = 5
n = len(sample)

z = (np.mean(sample) - pop_mean) / (pop_std / sqrt(n))
print("Z statistic:", z)

if abs(z) > 1.96:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

```python
# QUES:2Simulate random data to perform hypothesis testing and calculate the corresponding P-value using Python
import numpy as np
from scipy.stats import norm

sample = np.random.normal(55, 5, 40)
pop_mean = 50
pop_std = 5

z = (np.mean(sample) - pop_mean) / (pop_std / np.sqrt(len(sample)))
p_value = 2 * (1 - norm.cdf(abs(z)))

print("Z:", z)
print("P-value:", p_value)
```

```python
# QUES:3Implement a one-sample Z-test using Python to compare the sample mean with the population mean
import numpy as np
from scipy.stats import norm

sample = np.random.normal(102, 8, 60)
mu = 100
sigma = 8

z = (np.mean(sample) - mu) / (sigma / np.sqrt(len(sample)))
p = 1 - norm.cdf(z)

print("Z-score:", z)
print("P-value:", p)
```

```python
# QUES:9 Write a Python function to calculate the confidence interval for a sample mean and explain its significance
import numpy as np

def confidence_interval(data, z=1.96):
    mean = np.mean(data)
    std = np.std(data)
    n = len(data)
    return (mean - z*std/np.sqrt(n), mean + z*std/np.sqrt(n))

data = np.random.normal(100, 10, 50)
print(confidence_interval(data))
```

```python
# QUES:16 Implement an F-test using Python to compare the variances of two random samples
import numpy as np
from scipy.stats import f

x = np.random.normal(50, 5, 30)
y = np.random.normal(50, 10, 30)

f_stat = np.var(x, ddof=1) / np.var(y, ddof=1)
p_value = 1 - f.cdf(f_stat, 29, 29)

print("F-statistic:", f_stat)
print("P-value:", p_value)
```

```python
# QUES:24 Perform a hypothesis test for population variance using a Chi-square distribution and interpret the results
from scipy.stats import chi2
import numpy as np

data = np.random.normal(50, 5, 30)
sample_var = np.var(data, ddof=1)
pop_var = 25

chi = (len(data)-1) * sample_var / pop_var
p = 1 - chi2.cdf(chi, df=len(data)-1)

print("Chi-square:", chi)
print("P-value:", p)
```

```python
# QUES:25 Write a Python script to perform a Z-test for comparing proportions between two datasets or groups
import math

p1, n1 = 0.6, 100
p2, n2 = 0.5, 100

p_pool = (p1*n1 + p2*n2) / (n1+n2)
z = (p1 - p2) / math.sqrt(p_pool*(1-p_pool)*(1/n1 + 1/n2))

print("Z-score:", z)
```

```python
# QUES:26 Implement an F-test for comparing the variances of two datasets, then interpret and visualize the results
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import f

# Create two datasets
x = np.random.normal(50, 5, 30)
y = np.random.normal(50, 10, 30)

# F-test
f_stat = np.var(x, ddof=1) / np.var(y, ddof=1)
p_value = 1 - f.cdf(f_stat, 29, 29)

print("F-statistic:", f_stat)
print("P-value:", p_value)

# Visualization
variances = [np.var(x, ddof=1), np.var(y, ddof=1)]
plt.bar(["Sample 1", "Sample 2"], variances)
plt.title("Variance Comparison using F-test")
plt.ylabel("Variance")
plt.show()
```
