## Theoretical Questions (Clustering)
1. Q: What is unsupervised learning in the context of machine learning? A: Learning patterns from unlabeled data by discovering structure such as clusters, density, or low-dimensional representations.
2. Q: How does K-Means clustering algorithm work? A: It iteratively assigns points to the nearest centroid and updates centroids to the mean of assigned points until convergence.
3. Q: Explain the concept of a dendrogram in hierarchical clustering. A: A dendrogram is a tree diagram that shows the sequence of cluster merges (or splits) and the distances at which they occur.
4. Q: What is the main difference between K-Means and Hierarchical Clustering? A: K-Means requires a pre-set number of clusters and optimizes centroid variance, while hierarchical builds a tree of clusters without a fixed k.
5. Q: What are the advantages of DBSCAN over K-Means? A: DBSCAN finds arbitrarily shaped clusters, detects noise, and does not require specifying k.
6. Q: When would you use Silhouette Score in clustering? A: To evaluate cluster quality by measuring separation and cohesion when no labels exist.
7. Q: What are the limitations of Hierarchical Clustering? A: It can be computationally expensive for large datasets and is sensitive to noise and linkage choice.
8. Q: Why is feature scaling important in clustering algorithms like K-Means? A: K-Means uses distance; unscaled features can dominate and distort clusters.
9. Q: How does DBSCAN identify noise points? A: Points that are not density-reachable from any core point are labeled as noise.
10. Q: Define inertia in the context of K-Means. A: The sum of squared distances of points to their assigned centroids.
11. Q: What is the elbow method in K-Means clustering? A: A heuristic to choose k by finding where inertia stops decreasing sharply.
12. Q: Describe the concept of "density" in DBSCAN. A: Density is the number of points within an epsilon-radius neighborhood; dense regions form clusters.
13. Q: Can hierarchical clustering be used on categorical data? A: Yes, with an appropriate distance metric (e.g., Hamming) and linkage method.
14. Q: What does a negative Silhouette Score indicate? A: Points are likely assigned to the wrong cluster, overlapping with other clusters.
15. Q: Explain the term "linkage criteria" in hierarchical clustering. A: The rule for computing inter-cluster distance (single, complete, average, etc.).
16. Q: Why might K-Means perform poorly on data with varying cluster sizes or densities? A: It assumes spherical, equally sized clusters and minimizes variance around centroids.
17. Q: What are the core parameters in DBSCAN, and how do they influence clustering? A: `eps` sets neighborhood radius; `min_samples` sets density threshold for core points.
18. Q: How does K-Means++ improve upon standard K-Means initialization? A: It spreads initial centroids to reduce poor local minima and improve convergence.
19. Q: What is agglomerative clustering? A: A bottom-up hierarchical method that repeatedly merges the closest clusters.
20. Q: What makes Silhouette Score a better metric than just inertia for model evaluation? A: It balances cohesion and separation, while inertia only measures within-cluster compactness.


## Practical Tasks (Code)


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import make_blobs, make_moons, make_circles, load_iris, load_wine, load_breast_cancer, load_digits
from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.metrics import silhouette_score, silhouette_samples
```

### 21. make_blobs with 4 centers + K-Means


```python
X, _ = make_blobs(n_samples=400, centers=4, cluster_std=1.0, random_state=42)
km = KMeans(n_clusters=4, random_state=42, n_init=10)
labels = km.fit_predict(X)

plt.figure(figsize=(6,4))
plt.scatter(X[:,0], X[:,1], c=labels, cmap='tab10', s=20)
plt.scatter(km.cluster_centers_[:,0], km.cluster_centers_[:,1], c='black', s=120, marker='X')
plt.title('K-Means on 4-center blobs')
plt.show()
```

### 22. Iris + Agglomerative Clustering (3 clusters), first 10 labels


```python
iris = load_iris()
X = iris.data
agg = AgglomerativeClustering(n_clusters=3, linkage='ward')
labels = agg.fit_predict(X)
print('First 10 labels:', labels[:10])
```

### 23. make_moons + DBSCAN, highlight outliers


```python
X, _ = make_moons(n_samples=300, noise=0.08, random_state=42)
X_scaled = StandardScaler().fit_transform(X)

db = DBSCAN(eps=0.25, min_samples=5)
labels = db.fit_predict(X_scaled)

plt.figure(figsize=(6,4))
plt.scatter(X_scaled[:,0], X_scaled[:,1], c=labels, cmap='tab10', s=20)
plt.scatter(X_scaled[labels==-1,0], X_scaled[labels==-1,1], c='red', s=25, label='Noise')
plt.legend()
plt.title('DBSCAN on moons (noise highlighted)')
plt.show()
```

### 24. Wine + Standardize + K-Means, cluster sizes

### 28. Digits + PCA (2D) + K-Means

### 29. Silhouette scores for k=2..5 (make_blobs)


```python
X, _ = make_blobs(n_samples=400, centers=4, cluster_std=1.0, random_state=42)

scores = []
ks = [2, 3, 4, 5]
for k in ks:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X)
    scores.append(silhouette_score(X, labels))

plt.figure(figsize=(6,4))
plt.bar([str(k) for k in ks], scores)
plt.title('Silhouette Scores (k=2..5)')
plt.xlabel('k')
plt.ylabel('score')
plt.show()
```

### 30. Iris + hierarchical clustering dendrogram (average linkage)


```python
from scipy.cluster.hierarchy import dendrogram, linkage

iris = load_iris()
X = StandardScaler().fit_transform(iris.data)
Z = linkage(X, method='average')

plt.figure(figsize=(8,4))
dendrogram(Z, truncate_mode='lastp', p=12)
plt.title('Iris Dendrogram (average linkage)')
plt.xlabel('Cluster size')
plt.ylabel('Distance')
plt.show()
```

### 31. Overlapping blobs + K-Means + decision boundaries


```python
X, _ = make_blobs(n_samples=400, centers=3, cluster_std=2.2, random_state=42)
km = KMeans(n_clusters=3, random_state=42, n_init=10)
labels = km.fit_predict(X)

x_min, x_max = X[:,0].min() - 1, X[:,0].max() + 1
y_min, y_max = X[:,1].min() - 1, X[:,1].max() + 1
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 300), np.linspace(y_min, y_max, 300))
Z = km.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

plt.figure(figsize=(6,4))
plt.contourf(xx, yy, Z, alpha=0.25, cmap='tab10')
plt.scatter(X[:,0], X[:,1], c=labels, cmap='tab10', s=20)
plt.scatter(km.cluster_centers_[:,0], km.cluster_centers_[:,1], c='black', s=120, marker='X')
plt.title('K-Means decision boundaries')
plt.show()
```

### 34. Breast Cancer + inertia for k=2..6


```python
bc = load_breast_cancer()
X = StandardScaler().fit_transform(bc.data)

inertias = []
ks = [2, 3, 4, 5, 6]
for k in ks:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X)
    inertias.append(km.inertia_)

plt.figure(figsize=(6,4))
plt.plot(ks, inertias, marker='o')
plt.title('Inertia vs k (Breast Cancer)')
plt.xlabel('k')
plt.ylabel('inertia')
plt.show()
```
