1. What is a random variable in probability theory?

   Ans. A random variable is a function that assigns a numerical value to each outcome of a random experiment. It helps convert qualitative outcomes into quantitative form for analysis.

Example:

• Tossing a coin → X = number of heads

• Rolling a die → X = face value obtained

Random variables allow us to apply mathematical tools to probability problems.

2. What are the types of random variables?

   Ans. Types of Random Variables

(a) Discrete Random Variable

Takes a finite or countable number of values.

Examples:

• Number of students in a class

• Number of defective items

(b) Continuous Random Variable

Takes any value within a given range.

Examples:

• Height of a person

• Temperature

3. Difference between discrete and continous distributions?

  Ans. Discrete: $P(X=x)=p(x)$  

Continuous: $f(x)\ge0,\ \int_{-\infty}^{\infty} f(x)\,dx=1$

  
4. What are probability distribution functions (PDF)?

  Ans. For continuous random variables, the Probability Density Function (PDF) describes the likelihood of the variable taking a particular value.

Properties:

• f(x) ≥ 0

• Total area under curve = 1

Probability is calculated as area under the curve between two points.


5. Difference between CDF and PDF?

  Ans. PDF:

Shows probability density at a value.

CDF (Cumulative Distribution Function):

Shows accumulated probability up to that value.

$$
F(x) = P(X \le x) = \int_{-\infty}^{x} f(t)\,dt
$$


CDF is obtained by integrating PDF.

7. Key properties of a Bernoulli distribution?

  Ans.  A Bernoulli distribution represents a single trial with only two outcomes:

• Success (1)
• Failure (0)

Probability of success = p
Probability of failure = 1 − p

Mean = p
Variance = p(1−p)

Example: single coin toss


8. What is the binomial distribution and its use?

  Ans. Models number of successes in n independent Bernoulli trials.

Conditions:

• Fixed number of trials

• Two outcomes

• Constant probability

Formula:

$$
P(X = k) = \binom{n}{k} p^k (1 - p)^{n-k}
$$


Example: number of heads in 10 tosses.

22. What is expected value?

  Ans. Expected value is weighted average of outcomes.

Discrete:

$$
E(X) = \sum x P(x)
$$


Continuous:
$$
E(X) = \int_{-\infty}^{\infty} x f(x)\,dx
$$
