Ques:8 Difference between GET and POST HTTP methods ?

1. Purpose:

GET: Used to retrieve data from the server.

POST: Used to send or submit data to the server (like creating or updating something).

2️. Data location:

GET: Data is sent in the URL as query parameters. Example: /search?q=python

POST: Data is sent in the request body, not visible in the URL.

3️. Visibility:

GET: Data is visible in the browser URL.

POST: Data is hidden from the URL.

4️. Security:

GET: Less secure because URL parameters are visible and can be bookmarked.

POST: More secure because data is in the request body (not shown in URL).

5️. Length limit:

GET: Limited by URL length (~2000 characters).

POST: No practical limit; depends on server settings.

6️.  Idempotency:

GET: Yes – repeating the request does not change server state.

POST: No – repeating the request may create duplicates or change data.

7️.Caching:

GET: Can be cached by browsers and servers.

POST: Generally not cached.

8️.  Use case:

GET: Fetch/search data, like viewing a webpage or searching on Google.

POST: Submitting a form, sending login info, uploading files.

2️. What it includes:

RAML

API Blueprint

Ensures all routes behave the same way and return data in the expected format.

3️. Collaboration:
Front-end and back-end teams can work simultaneously because the API behavior is documented.

4️. Automatic documentation:
Tools like Swagger can generate interactive API docs from the specification.

5️. Error reduction:
Helps avoid confusion about input/output formats and prevents bugs.

Ques:7 How can you validate form data in Flask ?

```python
from flask import Flask, request, render_template_string

app = Flask(__name__)

@app.route('/', methods=['GET', 'POST'])
def form_example():
    error = ""
    if request.method == 'POST':
        # Get data from the form
        name = request.form.get('name')
        email = request.form.get('email')
        
        # Simple validation
        if not name or not email:
            error = "All fields are required!"
        elif "@" not in email:
            error = "Invalid email address!"
        else:
            return f"<h2>Form Submitted Successfully!</h2><p>Name: {name}</p><p>Email: {email}</p>"
    
    # HTML form with error display
    return render_template_string("""
    <html>
        <head><title>Flask Form Validation</title></head>
        <body>
            <h1>Fill the Form</h1>
            {% if error %}
                <p style="color:red;">{{ error }}</p>
            {% endif %}
            <form method="POST">
                Name: <input type="text" name="name"><br><br>
                Email: <input type="email" name="email"><br><br>
                <input type="submit" value="Submit">
            </form>
        </body>
    </html>
    """, error=error)

# Run Flask in Jupyter Notebook
app.run(debug=True, use_reloader=False)
```

Ques:8  How do you manage sessions in Flask ?

```python
from flask import Flask, session, redirect, url_for, request, render_template_string

app = Flask(__name__)

# Secret key is required to use sessions
app.secret_key = 'mysecretkey'

@app.route('/', methods=['GET', 'POST'])
def login():
    message = ""
    
    if request.method == 'POST':
        username = request.form.get('username')
        if username:
            # Store username in session
            session['username'] = username
            return redirect(url_for('welcome'))
        else:
            message = "Please enter a username!"
    
    return render_template_string("""
    <html>
        <head><title>Login Page</title></head>
        <body>
            <h1>Login</h1>
            {% if message %}
                <p style="color:red;">{{ message }}</p>
            {% endif %}
            <form method="POST">
                Username: <input type="text" name="username"><br><br>
                <input type="submit" value="Login">
            </form>
        </body>
    </html>
    """, message=message)

@app.route('/welcome')
def welcome():
    # Check if user is logged in
    if 'username' in session:
        username = session['username']
        return f"<h1>Welcome, {username}!</h1><a href='{url_for('logout')}'>Logout</a>"
    else:
        return redirect(url_for('login'))

@app.route('/logout')
def logout():
    # Remove user from session
    session.pop('username', None)
    return redirect(url_for('login'))

# Run Flask in Jupyter Notebook
app.run(debug=True, use_reloader=False)
```

Ques:12 How do you define a custom Jinja filter in Flask ?

```python
from flask import Flask, render_template_string

app = Flask(__name__)

# ----- Define a custom Jinja filter -----
def reverse_string(s):
    return s[::-1]

# Register the filter with Flask
app.jinja_env.filters['reverse'] = reverse_string

# ----- Route using the filter -----
@app.route('/')
def home():
    message = "Hello Flask"
    return render_template_string("""
    <html>
        <head><title>Custom Jinja Filter</title></head>
        <body>
            <h1>Original Message: {{ msg }}</h1>
            <h1>Reversed Message: {{ msg | reverse }}</h1>
        </body>
    </html>
    """, msg=message)

# Run Flask in Jupyter Notebook
app.run(debug=True, use_reloader=False)
```

Ques:13 How can you redirect with query parameters in Flask ?

```python
from flask import Flask, redirect, url_for, request

app = Flask(__name__)

@app.route('/')
def home():
    return "<h1>Home Page</h1><p>Go to <a href='/send'>Send Data</a></p>"

@app.route('/send')
def send_data():
    # Redirect to /receive with query parameters
    return redirect(url_for('receive', name='Chahat', age=20))

@app.route('/receive')
def receive():
    # Get query parameters
    name = request.args.get('name')
    age = request.args.get('age')
    return f"<h1>Received Data</h1><p>Name: {name}</p><p>Age: {age}</p>"

# Run Flask in Jupyter Notebook
app.run(debug=True, use_reloader=False)
```
