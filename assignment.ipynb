Ques:17  What is the difference between a class variable and an instance variable?

Ans:17 Definition:

Class Variable: A variable that is shared by all instances of a class.

Instance Variable: A variable that is unique to each object (instance) of a class.

Declaration:

Class Variable: Declared inside the class but outside any methods.

Instance Variable: Declared inside the constructor (__init__) or other instance methods using self.

Memory:

Class Variable: Stored once for the class.

Instance Variable: Stored separately for each object.

Access:

Class Variable: Can be accessed using ClassName.variable or self.variable.

Instance Variable: Accessed via self.variable only.

Effect of Change:

Class Variable: If modified via the class, changes reflect in all instances.

Instance Variable: Modifying it in one object does not affect other objects.

Ques:23 How does polymorphism work in Python with inheritanc ?

Ans:23efinition:
Polymorphism allows objects of different classes to be treated as objects of a common parent class. Methods in different classes can have the same name but behave differently depending on the object.

How it works with inheritance:

A child class can override a method from its parent class.

When the method is called on a child object, the child’s version is executed, not the parent’s.

```python
class Animal:
    def sound(self):
        print("Some generic sound")

class Dog(Animal):
    def sound(self):
        print("Bark")

class Cat(Animal):
    def sound(self):
        print("Meow")

# Polymorphism in action
animals = [Dog(), Cat(), Animal()]
for a in animals:
    a.sound()
```

 QUES:24 What is method chaining in Python OOP ?

 Ans:24 Definition:
Method chaining allows calling multiple methods in a single line by returning the object itself (self) from each method.

Purpose:

Makes code more concise and readable.

Common in builder patterns and fluent interfaces.

```python
class Person:
    def __init__(self, name):
        self.name = name
        self.age = 0

    def set_age(self, age):
        self.age = age
        return self  # Return self to enable chaining

    def greet(self):
        print(f"Hello, I am {self.name}, {self.age} years old")
        return self

# Method chaining
Person("Alice").set_age(25).greet()
```

 ans:25 Definition:
The __call__ method allows an object of a class to be called like a function.

Purpose:

Makes objects behave like functions.

Useful for creating function-like objects or callbacks.

5. Write a program to demonstrate encapsulation by creating a class BankAccount with private attributes
balance and methods to deposit, withdraw, and check balance.




```python
class BankAccount():
    def __init__(self,initial_balance=0):
        self.__balance=initial_balance
    
    def deposit(self,Amount):
        if Amount>0:
            self.__balance+=Amount
            print(f"Deposited:${Amount}")

        else:
            print("deposit amount is invalid")

    def withdraw(self,Amount):
        if Amount > 0:
            if Amount <= self.__balance:
                self.__balance-=Amount
                print(f"withdrawn:${Amount}")
            else:
                print("Insufficient balance")
        else:
            print("invalid")

    def get_balance(self):
        return self.__balance

c=BankAccount()
c.deposit(100000)
c.withdraw(1000)
c.get_balance()

```

10. Demonstrate operator overloading by creating a class Vector and overriding the add method to add two
vectors

```python
class vector:
    def __init__(self,x,y):
        self.x=x
        self.y=y

    def __add__(self,other):
        if isinstance(other,vector):
            return vector(self.x+other.x,self.y+other.y)
        else:
            raise ValueError("Can only add another Vector")
        
    def __str__(self):
        return f"Vector({self.x},{self.y})"
v1=vector(2,3)
v2=vector(4,5)
v3=v1+v2
print(v3)
```

```python
class  Student():
    def __init__(self,name,grades):
        self.name=name
        self.grades=grades

    def average_grade(self):
        if len(self.grades)==0:
            return 0
        else:
            return sum(self.grades)/len(self.grades)
        
    def __str__(self):
        return f"Student:{self.name},Average Grade:{self.average_grade():.2f}"
    
s1=Student("Chahat",[85,90,78,92])
s1.average_grade()
print(s1)
```

14. Create a class Employee with a method calculate_salary() that computes the salary based on hours worked
and hourly rate. Create a derived class Manager that adds a bonus to the salary

```python
class Employee():
    def __init__(self,hours=0,hourly_rate=100):
        self.hours=hours
        self.hourly_rate=hourly_rate
    
    def calculate_salary(self):
        return self.hours*self.hourly_rate
#derived class
class Manager(Employee):
    def __init__(self,hours=0,hourly_rate=100,bonous=0):
        super().__init__(hours, hourly_rate)
        self.bonous=bonous

    def calculate_salary(self):
       base_salary=super().calculate_salary()
       return base_salary+self.bonous
    
e1=Employee(hours=40,hourly_rate=100)
e1.calculate_salary()
m1=Manager(hours=20,hourly_rate=90,bonous=2000)
m1.calculate_salary()
```
