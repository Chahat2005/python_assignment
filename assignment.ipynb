QUES:2 How does broadcasting work in NumPy?

ANS:2 Broadcasting allows arrays of different shapes to be combined in operations.

Rules: dimensions are compatible if they are equal or one of them is 1.

Example:

```python
import numpy as np
a = np.array([1,2,3])
b = np.array([[1],[2],[3]])
print(a + b)  # a is broadcasted to match b's shape
```

QUE:4 Explain the use of the groupby() method in Pandas

ANS:4Used to split data into groups based on column(s), then aggregate, transform, or filter.

Example: Sum sales by region:


```python
import pandas as pd

# Sample dataset
data = {
    'region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West'],
    'sales': [250, 150, 200, 300, 100, 200, 150, 250]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display DataFrame
print("Original DataFrame:")
print(df)

# Group by region and sum sales
grouped = df.groupby('region')['sales'].sum()
print("\nSales sum by region:")
print(grouped)
```

QUES:6-Differences between NumPy arrays and Python lists

ANS:6 Data Type:

NumPy arrays are homogeneous, meaning all elements must be of the same data type (e.g., all integers or all floats).

Python lists are heterogeneous, so they can contain elements of different types (e.g., integers, strings, floats together).

Size:

NumPy arrays have a fixed size once created.

Python lists are dynamic, allowing elements to be added or removed.

Performance:

NumPy arrays are faster for numerical computations due to vectorized operations.

Python lists are slower because element-wise operations require explicit loops.

Memory Usage:

NumPy arrays use less memory because data is stored in contiguous memory blocks.

Python lists use more memory due to storing additional metadata for each element.

Operations:

NumPy arrays support element-wise arithmetic and mathematical operations directly.

Python lists require loops or list comprehensions for element-wise operations.

Functionality:

NumPy provides advanced features like linear algebra, statistical functions, random sampling, reshaping arrays, and broadcasting.

Python lists have only basic functionality for storing and accessing data.

Indexing and Slicing:

NumPy arrays support advanced slicing and multi-dimensional indexing.

Python lists only support standard indexing and slicing.

Integration:

NumPy arrays integrate easily with scientific libraries like Pandas, Matplotlib, and SciPy.

Python lists have limited integration for numerical or scientific computing.

QUES:17 Explain the difference between apply() and map() in Pandas?

ANS:17 The map() function is used primarily with a Pandas Series to apply a function element-wise, transforming each individual value. In contrast, apply() is more flexible: it can be used on both Series and DataFrames, and can apply functions row-wise, column-wise, or on the entire DataFrame. While map() returns a Series, apply() can return either a Series or a DataFrame depending on the operation. Essentially, map() is simpler and used for element-wise transformations, whereas apply() handles more complex operations, including aggregations and transformations across rows or columns.

QUES:20 Why is NumPy’s array slicing faster than Python’s list slicing ?

ANS:20 NumPy arrays are stored as contiguous blocks of memory, unlike Python lists which store references to objects. This allows NumPy to perform operations at the C-level, minimizing overhead. NumPy slicing uses vectorized operations, which avoid Python-level loops and type checking for each element, making array operations significantly faster than equivalent operations on Python lists.

Ques:13  Create a 3D scatter plot using Plotly.

```python
import pandas as pd
import plotly.graph_objects as go

# Sample DataFrame
data = {
    'Geography': [1, 2, 3, 4, 5],
    'CreditScore': [600, 650, 700, 720, 680],
    'Age': [25, 30, 40, 35, 28]
}

df = pd.DataFrame(data)

# Create a 3D scatter plot
fig = go.Figure()

fig.add_trace(go.Scatter3d(
    x=df['Geography'],
    y=df['CreditScore'],
    z=df['Age'],
    mode='markers',
    marker=dict(size=5, color=df['CreditScore'], colorscale='Viridis', opacity=0.8)
))

# Set plot layout
fig.update_layout(title='3D Scatter Plot', scene=dict(
    xaxis_title='Geography',
    yaxis_title='Credit Score',
    zaxis_title='Age'
))

# Display the figure
fig.show()
```
