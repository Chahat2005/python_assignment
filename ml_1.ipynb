## Regression Assignment - All Questions with Answers

Below are all assignment questions answered in sequence.

### Q1. What is Simple Linear Regression?
**Answer:** Simple Linear Regression (SLR) is a method to model the relationship between one independent variable (X) and one dependent variable (Y) using a straight line: Y = mX + c.

### Q2. What are the key assumptions of Simple Linear Regression?
**Answer:** Main assumptions are linearity, independence of observations, constant variance of errors (homoscedasticity), normally distributed residuals, and no influential extreme outliers.

### Q3. What does the coefficient m represent in the equation Y = mX + c?
**Answer:** m is the slope. It tells how much Y changes for a one-unit increase in X.

### Q4. What does the intercept c represent in the equation Y = mX + c?
**Answer:** c is the intercept. It is the predicted value of Y when X = 0.

### Q5. How do we calculate the slope m in Simple Linear Regression?
**Answer:** m = sum((Xi - X_bar)(Yi - Y_bar)) / sum((Xi - X_bar)^2). It gives the best-fit slope under least squares.

### Q6. What is the purpose of the least squares method in Simple Linear Regression?
**Answer:** Least squares finds the line that minimizes the sum of squared residuals (prediction errors), giving the best overall fit to data.

### Q7. How is the coefficient of determination (R^2) interpreted in Simple Linear Regression?
**Answer:** R^2 is the proportion of variance in Y explained by X. For example, R^2 = 0.80 means 80% of Y variability is explained by the model.

### Q8. What is Multiple Linear Regression?
**Answer:** Multiple Linear Regression (MLR) models one dependent variable using two or more independent variables: Y = b0 + b1X1 + b2X2 + ... + bnXn.

### Q9. What is the main difference between Simple and Multiple Linear Regression?
**Answer:** SLR uses one predictor, while MLR uses multiple predictors.

### Q10. What are the key assumptions of Multiple Linear Regression?
**Answer:** Linearity, independence of errors, homoscedasticity, normality of residuals, no perfect multicollinearity, and correct model specification.

### Q11. What is heteroscedasticity, and how does it affect the results of a Multiple Linear Regression model?
**Answer:** Heteroscedasticity means non-constant error variance. It can make standard errors unreliable, leading to incorrect p-values and confidence intervals.

### Q12. How can you improve a Multiple Linear Regression model with high multicollinearity?
**Answer:** Remove or combine highly correlated variables, use VIF checks, apply feature selection, use PCA, or use regularization methods like Ridge/Lasso.

### Q13. What are some common techniques for transforming categorical variables for use in regression models?
**Answer:** Common methods are one-hot encoding, ordinal encoding (for ordered categories), binary encoding, and target/frequency encoding (with care).

### Q14. What is the role of interaction terms in Multiple Linear Regression?
**Answer:** Interaction terms model combined effects where the impact of one variable depends on another, e.g., b3(X1*X2).

### Q15. How can the interpretation of intercept differ between Simple and Multiple Linear Regression?
**Answer:** In SLR, intercept is Y at X=0. In MLR, intercept is Y when all predictors are 0, which may or may not be practically meaningful.

### Q16. What is the significance of the slope in regression analysis, and how does it affect predictions?
**Answer:** Each slope coefficient shows expected change in Y for a one-unit increase in that predictor (holding others constant in MLR). Predictions change directly with slope size and sign.

### Q17. How does the intercept in a regression model provide context for the relationship between variables?
**Answer:** The intercept sets the baseline prediction level. It anchors the regression line or hyperplane before adding predictor effects.

### Q18. What are the limitations of using R^2 as a sole measure of model performance?
**Answer:** R^2 does not confirm causality, can increase with irrelevant features, does not reveal bias patterns, and does not measure prediction error on new data.

### Q19. How would you interpret a large standard error for a regression coefficient?
**Answer:** A large standard error means the coefficient estimate is unstable/uncertain, often due to noise, small sample size, or multicollinearity.

### Q20. How can heteroscedasticity be identified in residual plots, and why is it important to address it?
**Answer:** In residual vs fitted plots, a funnel or changing spread indicates heteroscedasticity. It must be addressed to get valid inference and more reliable uncertainty estimates.

### Q21. What does it mean if a Multiple Linear Regression model has a high R^2 but low adjusted R^2?
**Answer:** It suggests overfitting: added predictors inflate R^2 but do not contribute meaningful explanatory power after complexity penalty.

### Q22. Why is it important to scale variables in Multiple Linear Regression?
**Answer:** Scaling improves numerical stability, helps compare coefficient magnitudes, and is important for regularized regression methods.

### Q23. What is polynomial regression?
**Answer:** Polynomial regression models non-linear relationships by adding powers of predictors (X^2, X^3, etc.) while remaining linear in coefficients.

### Q24. How does polynomial regression differ from linear regression?
**Answer:** Linear regression fits a straight-line relationship in original feature space, while polynomial regression fits curved patterns using transformed polynomial features.

### Q25. When is polynomial regression used?
**Answer:** It is used when data shows a curved trend and a straight line underfits the relationship.

### Q26. What is the general equation for polynomial regression?
**Answer:** For one variable: Y = b0 + b1X + b2X^2 + ... + bnX^n + e.

### Q27. Can polynomial regression be applied to multiple variables?
**Answer:** Yes. Polynomial and interaction terms can be created for multiple predictors, e.g., X1^2, X2^2, X1X2.

### Q28. What are the limitations of polynomial regression?
**Answer:** High degrees can overfit, extrapolation can be unstable, coefficients become hard to interpret, and multicollinearity may increase.

### Q29. What methods can be used to evaluate model fit when selecting the degree of a polynomial?
**Answer:** Use train/validation split or cross-validation, compare RMSE/MAE/R^2, inspect adjusted R^2, and use AIC/BIC where appropriate.

### Q30. Why is visualization important in polynomial regression?
**Answer:** Visualization helps verify curvature, compare fitted curves across degrees, and detect underfitting or overfitting.

### Q31. How is polynomial regression implemented in Python?
**Answer:** Typical workflow: create polynomial features (sklearn.preprocessing.PolynomialFeatures), fit LinearRegression, and evaluate with metrics/cross-validation.
